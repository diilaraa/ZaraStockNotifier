```python
import asyncio
from playwright.async_api import async_playwright

PRODUCT_URL = "https://www.zara.com/tr/tr/genis-cepli-pantolon-p03644180.html?v1=445273608&v2=2420795"  # Ürün URL'sini buraya ekle
CHECK_INTERVAL = 60  # saniye cinsinden kontrol aralığı

async def check_stock(page, url):
    await page.goto(url, timeout=60000)
    try:
        add_to_cart = await page.query_selector('button[data-qa-action="add-to-cart"]')
        if add_to_cart and await add_to_cart.is_enabled():
            return True
    except:
        pass
    return False

async def main():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=False)
        page = await browser.new_page()

        while True:
            print("Stok kontrolü yapılıyor...")
            in_stock = await check_stock(page, PRODUCT_URL)
            if in_stock:
                print("🎉 Ürün stokta! Hemen kontrol et!")
                send_telegram_message("🎯 Zara ürünü stokta! Hemen sepete ekle veya incele: " + PRODUCT_URL)
                break
            else:
                print("Ürün stokta değil. Bekleniyor...")
            await asyncio.sleep(CHECK_INTERVAL)

        await browser.close()

# Eğer Jupyter'deysen:
await main()
```
